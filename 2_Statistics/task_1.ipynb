# STATISTICAL TESTS Task

```python
import numpy as np
from scipy import stats
from statsmodels.stats import proportion

# import scipy
```

```python
def generate_sample(mean, sd, n, seed=None):
    rng = np.random.default_rng(seed)

    arr = rng.normal(size=n)

    # Standardize
    arr = (arr - arr.mean()) / arr.std(ddof=1)

    # Rescale
    arr = arr * sd + mean

    return arr
```

## Question 1
#### The historical average score in a certification exam is 65. A training institute introduces a new course. A random sample of 50 students trained under the new course has a mean score of 68 and a standard deviation of 10. At the 5% level of significance, test whether the new course improves performance.

1. Noted Points: 
    - Population Mean = 65
    - Sample Size = 50
    - Sample Mean = 68
    - Sample SD = 10
    - Significance Level = 5% = 0.05
2. Hypothesis
    - H0 : mean = 65
    - Ha : mean > 65
3. Test : One Sample t-test Right Tailed


```python
student_score = generate_sample(mean=68, sd=10, n=50, seed=42)

t_stats, p_value = stats.ttest_1samp(a=student_score, popmean=65, alternative="greater")

print(f"t-Statistics: {t_stats}")
print(f"p-Value: {p_value}")

if p_value < 0.05:
    print("Reject H0.")
else:
    print("Do no Reject H0")
```

1. Results:
   - t-statistic = 2.1213
   - p-value = 0.0194

1. Decision: 
    - p-value (0.0194) < significant level (0.05), reject H0

2. Conclusion:
    - At 5% level of significance level, we reject the null Hypothesis. There is sufficient evidence to conclude that the new course improves student peformance.

## Question 5
#### Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10. At the 5% level of significance, test whether there is a difference in average scores between the two methods.



1. Noted Points:
    - Method A:
        - Sample Size = 30
        - Sample Mean = 75
        - Sample SD = 12
    - Method B:
        - Sample Size = 28
        - Sample Mean = 70
        - Sample SD = 10
    - Significance leve = 5% = 0.05

2. Hypothesis:
    - H0 = mean(A) = mean(B)
    - Ha = mean(A) != mean(B)

3. Test:
    - Two-Sample t-test Two Tailed ( Welch's t-test)

```python
method_a = generate_sample(mean=75, sd=12, n=30, seed=42)
method_b = generate_sample(mean=70, sd=10, n=28, seed=123)

t_stats, p_value = stats.ttest_ind(
    a=method_a, b=method_b, equal_var=False, alternative="two-sided"
)

print(f"t-Statistics: {t_stats}")
print(f"p-Value: {p_value}")

if p_value < 0.05:
    print("Reject H0.")
else:
    print("Do no Reject H0")
```

1. Result:
    - t-statistics = 1.7281
    - p-value = 0.0895

2. Decision:
    - p-value (0.0895) > Significance Level (0.05), do not reject H0

3. Conclusion:
    - At the 5% significance level, we do not reject null hypothesis. There is insufficient evidence to conclude that the average scores differ between method A and method B.

        - Sample Size = 25
        - Sample Mean = 80
        - Sample SD = 8
    - Program B:
        - Sample Size = 30
        - Sample Mean = 75
        - Sample SD = 7
    - Program C:
        - Sample Size = 28
        - Sample Mean = 82
        - Sample SD = 6
    - Significance leve = 5% = 0.05

2. Hypothesis:
    - H0 = mean(A) = mean(B) = mean(B)
    - Ha = At least one program mean differs

3. Test:
    - One-Way ANOVA (Welch's One-Way ANOVA)

```python
program_a = generate_sample(mean=80, sd=8, n=25, seed=42)
program_b = generate_sample(mean=75, sd=7, n=30, seed=123)
program_c = generate_sample(mean=82, sd=6, n=28, seed=456)


f_stats, p_value = stats.f_oneway(program_a, program_b, program_c, equal_var=False)

print(f"f-statistic: {f_stats}")
print(f"p-value: {p_value}")

if p_value < 0.05:
    print("Reject H0.")
else:
    print("Do Not Reject H0.")
```

1. Result:
    - f-statistic = 8.4620
    - p-value = 0.0006

2. Decision:
    - p-value(0.0006) < significant level (0.05), reject H0

3. Conclustion
    - At 5% significance leavel, we reject the null hypothesis. There is sufficient evidence to conclude that at least one training program has a differnct mean productivity scroe.

## Question 10
#### 10.	A company wants to test whether gender and product preference are independent. The survey results are as follows:
- #### Male: 60 prefer Product A, 40 prefer Product B
- #### Female: 50 prefer Product A, 50 prefer Product B 
#### At the 5% level of significance, test for independence.

1. Noted Points
    - Male:
        - Product A = 60
        - product B = 40
    - Femal:
        - Product A = 50
        - product B = 50
    - significance level = 5% = 0.05

2. Hypothesis
    - H0: Gender and Product preference are independent
    - Ha: Gender and Product preference are not independent

3. Test: Chi- Square Test of Independence

```python
table = [
    [60, 40],
    [50, 50]
]

chi2, p_value, dof, expected = stats.chi2_contingency(observed=table)


print(f"chi-square Statistic: {chi2}")
print(f"p-value: {p_value}")
print(f"Expected: \n{expected}")

if p_value < 0.05:
    print("Reject H0.")
else:
    print("Do Not Reject H0.")
```

1. Result:
    - Chi-square Statistic = 1.6363
    - p-value = 0.2008

2. Decision:
    - p_value (0.2008) > significant level (0.05), do not reject Null Hypothesis

3. At 5% level of significance, we do not reject null hypothesis. There is insufficient evidence to conclude that gender and product preference are associated. Therefore, the data suggest that gendera and product preference are independent.

## Question 12
#### A dice manufacturer claims that a die is fair. The die is rolled 120 times, and the observed frequencies are:
- #### Face 1: 18
- #### Face 2: 22
- #### Face 3: 20
- #### Face 4: 25
- #### Face 5: 17
- #### Face 6: 18
#### At the 5% level of significance, test whether the die is fair.

```python
# Expected frequency for each face:
120/6
```

1. Noted Points:
    - Observed Frequencies:
        - Face 1 = 18
        - Face 2 = 22
        - Face 3 = 20
        - Face 4 = 25
        - Face 5 = 17
        - Face 6 = 18
    - Total Rolls = 120
    - Expected Frequenies = [20, 20, 20, 20, 20, 20]
    - Significance Level = 5% = 0.05

2. Hypothesis
    - H0: The die is fair
    - Ha: The die is not fair
3. Test: Chi-Square Goodness-of-Fit Test

```python
observed = [18, 22, 20, 25, 17, 18]
expected = [20] * 6

chi2, p_value  = stats.chisquare(f_obs=observed, f_exp=expected)

print(f"Chi-square Statistic: {chi2}")
print(f"P-value: {p_value}")

if p_value < 0.05:
    print("Reject H0.")
else:
    print("Do Not Reject H0.")
```

1. Result:
    - Chi-square Statistic = 2.30
    - p-value ≈ 0.807

2. Decision:
    - p-value (0.807) > significant level (0.05), do not reject H0.

3. Conclusion:
    - At the 5% significance level, we do not reject the null hypothesis. There is insufficient evidence to conclude that the die is unfair. The observed frequencies are consistent with a fair die.
