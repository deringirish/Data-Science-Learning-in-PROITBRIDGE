#### 1. A company claims that the average time taken to resolve customer tickets is 30 minutes. A random sample of 40 tickets shows a mean resolution time of 33 minutes. At the 5% level of significance, test whether the average resolution time has increased.

```python
# H0 -> Current Process / Statement of no change - There is no  significant difference in the average resolution time
# H1 -> New Process / Statement of change - There is a significant difference in the average resolution time
```

```python
from scipy import stats
import numpy as np
import pandas as pd
```

```python
observation_40_tickets = np.random.normal(loc=33, scale=5.0, size=40).astype("int")
print(observation_40_tickets)
observation_40_tickets.mean()
```

```python
t_stats, p_value = stats.ttest_1samp(a=observation_40_tickets, popmean=30, alternative="greater")
```

```python
t_stats, p_value
```

```python
if p_value < 0.05:
    print("We Reject Null Hypothesis H0")
else:
    print("We do no Reject Null Hypothesis H0")
```

### Conclusion
#### At 5% Level of Significance we can reject the null hypothesis and claim that the average resolution time is increase
