## Level - 3: Give Multiple Conditions
1. user should enter the number between 1 - 100. If it's entered wrong, the program should say, "Invalid Number"
2. If the user guesed number > secret number, then program should say, "Your number is greater than the secrete number"
2. If the user guesed number < secret number, then program should say, "Your number is less than the secrete number"

```python
print("Welcome to Number Gussing Game!!!")
print("Guess the Number between 1 t0 100")

secret_number = 98

for i in range(7):
    user_input = int(input("Enter the Number: "))
    if user_input < 1 or user_input > 100:
        print("Invalid Number! Please enter the number between 1 - 100")
    elif user_input < secret_number:
        print("Your Number is Lesser")
    elif user_input > secret_number:
        print("Your Number is Higher")
    else:
        print("Gussed Number is Correct")
        print(f"Congradulations, you have guessed the number {secret_number} in {i+1} attemptes")
        break

```
