## 1. List

```python
cart = ["Milk", "Bread", "Eggs"]
print(f"Cart: {cart}")
```

#### 1.1. append()

```python
print(f"Cart: {cart}")
cart.append("Cheese")
print(f"Cart: {cart}")
```

#### 1.2 insert()

```python
print(f"Cart: {cart}")
cart.insert(3, "Juice")
print(f"Cart: {cart}")
```

#### 1.3 extend()

```python
print(f"Cart: {cart}")
cart.extend(["rice", "carrot"])
print(f"Cart: {cart}")
```

#### 1.4 count()

```python
cart.extend(["Bread", "Eggs"])
print(f"Cart: {cart}")
count = cart.count('Bread')
print(f"Count of Bread: {count}")
```

#### 1.5 index()

```python
print(f"Cart: {cart}")
index = cart.index("Eggs")
print(f"Index of Eggs: {index}")
```

#### 1.6 remove()

```python
print(f"Cart: {cart}")
cart.remove("Milk")
print(f"Cart: {cart}")
```

#### 1.7 pop()

```python
print(f"Cart: {cart}")
popped_item = cart.pop()
print(f"Popped Item: {popped_item}")
print(f"Cart: {cart}")
```

#### 1.8 sort()

```python
print(f"Cart: {cart}")
cart.sort()
print(f"Cart: {cart}")
```

#### 1.9 reverse()

```python
print(f"Cart: {cart}")
cart.reverse()
print(f"Cart: {cart}")
```

```python
copy_cart = cart.copy()
print(f"Cart: {cart}")
print(f"Cart: {copy_cart}")
```

```python
print(f"Cart: {cart}")
cart.clear()
print(f"Cart: {cart}")
```

## 2. Set

```python
coding_club = {"Derin", "Girish", "Adarsh", "Aditya"}
music_club = {"Adarsh", "Monish", "Girish"}
print(f"Coding Club: {coding_club}")
print(f"Music Club: {music_club}")
```

#### 2.1 add()

```python
print(f"Coding Club before adding: {coding_club}")
coding_club.add("Deena")
print(f"Coding Club after adding: {coding_club}")
```

#### 2.2 update()

```python
print(f"Coding Club before update: {coding_club}")
coding_club.update({"Harsha", "Grunath"})
print(f"Coding Club after update: {coding_club}")
```

#### 2.3 union()

```python
print(f"Coding Club: {coding_club}")
print(f"Music Club: {music_club}")
all_student = coding_club.union(music_club)
print(f"All Student: {all_student}")
```

#### 2.4 intersection()

```python
print(f"Coding Club: {coding_club}")
print(f"Music Club: {music_club}")
common_student = coding_club.intersection(music_club)
print(f"Common Student: {common_student}")
```

#### 2.5 intersection_update()

```python
all_student_1 = all_student.copy()
print(f"All Student Before Insersection Update: {all_student_1}")
print(f"Music Club: {music_club}")
all_student_1.intersection_update(music_club)
print(f"All Student After Insersection Update: {all_student_1}")
```

#### 2.6 difference()

```python
print(f"Coding Club: {coding_club}")
print(f"Music club: {music_club}")
only_coding_club = coding_club.difference(music_club)
print(f"Only Coding Club: {only_coding_club}")
```

#### 2.7 difference_update()

```python
all_student_1 = all_student.copy()
print(f"All Student before Difference Update: {all_student_1}")
print(f"Music Club: {music_club}")
all_student_1.difference_update(music_club)
print(f"All Student after Difference Update: {all_student_1}")

```

#### 2.8 symmetric_difference()

```python
print(f"Coding Club: {coding_club}")
print(f"Music Club: {music_club}")
unique_student = coding_club.symmetric_difference(music_club)
print(f"Unique Student: {unique_student}")
```

#### 2.9 summertic_difference_update()

```python
coding_club_1 = coding_club.copy()
print(f"Coding Club before symmetric differnce update : {coding_club_1}")
coding_club_1.symmetric_difference_update(music_club)
print(f"Music Club: {music_club}")
print(f"Coding Club after symmetric differnce update : {coding_club_1}")
```

#### 2.10 isdisjoint()

```python
sports_club = {"nandha", "rishi"}
print(f"Coding Club: {coding_club}")
print(f"Music Club: {music_club}")
print(f"Sports Club: {sports_club}")

print(coding_club.isdisjoint(music_club))

print(coding_club.isdisjoint(sports_club))
```

#### 2.11 issubset()

```python
ds_club = {"Derin", "Girish", "Aditya"}
print(f"Coding Club: {coding_club}")
print(f"DS Club: {ds_club}")
print(f"Sports Club: {sports_club}")

print(ds_club.issubset(coding_club))
print(sports_club.issubset(coding_club))
```

#### 2.12 issuperset()

```python
print(f"Coding Club: {coding_club}")
print(f"DS Club: {ds_club}")
print(f"Sports Club: {sports_club}")

print(coding_club.issuperset(ds_club))
print(coding_club.issuperset(sports_club))
```

#### 2.13 remove()

```python
print(f"Coding Club: {coding_club}")
coding_club.remove("Aditya")
print(f"Coding Club after remove: {coding_club}")
```

#### 2.14 discard()

```python
print(f"Coding Club: {coding_club}")
coding_club.discard("Harsha")
print(f"Coding Club after discard: {coding_club}")
```

#### 2.15 pop()

```python
print(f"coding_club: {coding_club}")
popped_stuend = coding_club.pop()
print(f"Popped Student: {popped_stuend}")
```

#### 2.16 copy()

```python
copy_music_club = music_club.copy()
print(f"Music Club: {music_club}")
print(f"Music Club Copy: {copy_music_club}")
```

```python
print(f"Music club: {music_club}")
music_club.clear()
print(f"Music Club: {music_club}")
```

## 4. Dictionary

```python
student_data = {
    "name" : "Derin",
    "age" : 22,
    "course" : "CSE"
}

print(f"Student Data : {student_data}")
```

#### 4.1 get()

```python
name = student_data.get("name")
age = student_data.get("age")
print(f"Name: {name}")
print(f"Age: {age}")
```

#### 4.2 keys()

```python
print(f"Keys: {student_data.keys()}")
```

#### 4.3 values()

```python
print(f"Values: {student_data.values()}")
```

#### 4.4 items()

```python
print(f"Items: {student_data.items()}")
```

#### 4.5 update()

```python
print(f"Student Data: {student_data}")
student_data.update({"city" : "Bangalore"})
print(f"Student Data: {student_data}")
```

#### 4.6 setdefault()

```python
print(f"Student Data: {student_data}")
student_data.setdefault("grade", "A")
print(f"Student Data: {student_data}")
student_data.setdefault("city", "Chennai")
print(f"Student Data: {student_data}")
```

#### 4.7 pop()

```python
print(f"Student Student: {student_data}")
poped_item = student_data.pop("course")
print(f"Poped Value: {poped_item}")
print(f"Student Student: {student_data}")
```

#### 4.8 popitem()

```python
print(f"Student Data: {student_data}")
poped_item = student_data.popitem()
print(f"Poped Value: {poped_item}")
print(f"Student Data: {student_data}")
```

#### 4.9 copy()

```python
student_data_copy = student_data.copy()
print(f"Student Data: {student_data}")
print(f"Student Data Copy: {student_data_copy}")
```

#### 4.10 fromkeys()

```python
marks_tuple = ("tamil", "english", "math", "science", "social")
mark = 35
marks_dictionary =  dict.fromkeys(marks_tuple, mark)
print(f"Marks: {marks_dictionary}")
```

#### 4.11 clear()

```python
print(f"Student Data: {student_data}")
student_data.clear()
print(f"Student Data: {student_data}")
```
