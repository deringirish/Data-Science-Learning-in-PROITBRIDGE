## 2. Set

- 1D
- Hetrogenous 
- Won't allow duplicate Values
- indexing & slicing not possible
- maintain Storted order
- mutable

```python
empyt_cart_2 = set()
```

```python
empyt_cart_2
```

```python
type(empyt_cart_2)
```

```python
items_to_buy_2 = {"apple", "orange", "grapes", "pomogranate", "apple", 6.76, 8, 9}
items_to_buy_2
```

```python
type(items_to_buy_2)
```

```python
items_to_buy_2[1]
```

```python
items_to_buy_2[2] = "watermeolon"
```

```python
items_to_buy_2.add("WaterMelon")
items_to_buy_2
```

```python
items_to_buy_2.discard("apple")
items_to_buy_2
```

```python
empyt_cart_3 = ()
```

```python
empyt_cart_3
```

```python
type(empyt_cart_3)
```

```python
items_to_buy_3 = ("apple", "orange", "grapes", "pomogranate", "apple", 6.76, 8, 9)
items_to_buy_3
```

```python
items_to_buy_3[2]
```
