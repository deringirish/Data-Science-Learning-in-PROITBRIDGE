## Create 3 DataFrames using Dictionaries

---

```python
import pandas as pd
```

## 1. DataFrame 1

```python
student_data = {
    "USN" : [101, 102, 103, 104, 105],
    "Name" : ["Derin", "Girish", "Monish", "Deena", "Aditya"],
    "Age" : [22, 22, 21, 23, 21]
}
student_data
```

```python
student_data_frame = pd.DataFrame(data=student_data)
student_data_frame
```

---

## 2. DatFrame 2

```python
employee_data = [
    {"Emp_ID" : 101, "Name" : "Derin", "Salary" : 75000},
    {"Emp_ID" : 102, "Name" : "Girish", "Salary" : 71000},
    {"Emp_ID" : 103, "Name" : "Adarsh", "Salary" : 45000},
    {"Emp_ID" : 104, "Name" : "Nandha", "Salary" : 97000},
    {"Emp_ID" : 105, "Name" : "Yeswanth", "Salary" : 24000},
]
employee_data
```

```python
employee_data_frame = pd.DataFrame(data=employee_data)
employee_data_frame
```

---

## 3. DataFrame 3

```python
product_data = {
    "P_101" : {"Product Name" : "Laptop", "Price" : 75000},
    "P_102" : {"Product Name" : "Mobile", "Price" : 20000},
    "P_103" : {"Product Name" : "Tablet", "Price" : 38000},
}
product_data
```

```python
product_data_frame = pd.DataFrame.from_dict(data=product_data, orient='index')
product_data_frame
```
