# Data Visualization Introduction

#### Why Data Visualization
1. Easy Understanding of Data
2. It saves more time because we can quickly analyze the data
3. For story telling
4. Data Quality Check

#### Python Based Viz Libraries
1. matplotlib - more custamizable on my graph/chart
2. Seaborn - For beautification purpose
3. Plotly - to create interactive charts

## On what basis should i need to choose between the charts
1. **Univariate Analysis** - to understand the quality of one feature -> Countplot, Histogram, Boxplot, PieChart
2. **Bivariate Analysis** - to understand the quality/relationship between two features -> Barplot, Scatterplot
3. **Multivariate Analysis** - to understand the relationship between more than 2 featues -> pairplot
- lineplot -> for time series data
