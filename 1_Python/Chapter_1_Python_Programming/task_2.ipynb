#### 1. while loop

```python
i = 1
while i <= 10:
  print(i)  
  i = i + 1
```

```python
original_password = "Derin_123_@#"
while True:
  password = input("Enter the password: ")
  if password == original_password:
    break
```

#### 2. for loop

```python
name = "Derin Girish"
for i in name:
  print(i)
```

```python
table_name = int(input("Enter table Name: "))
for i in range(1, 11):
  print(f"{i} x {table_name} = {i * table_name}")
```

#### 3.	if else

```python
age = int(input("Enter you age: "))
if age >= 18:
  print("Adult")
else:
  print("Child")
```

```python
number = int(input())
if number % 2 == 0:
  print("Even")
else:
  print("Odd")
```

#### 4.	if elif else

```python
mark = int(input("Enter the mark: "))
if 0 > mark or mark > 100:
  print("Invalid Mark")
elif mark >= 90:
  print("Excellent")
elif mark >= 70:
  print("Good")
elif mark >= 50:
  print("Average")
elif mark >= 30:
  print("Poor")
else:
  print("Fail")
```

```python
signal = input("Enter signal Color: ").lower()
if signal == "red":
  print("Stop")
elif signal == "yellow":
  print("Wait")
elif signal == "green":
  print("Go")
else:
  print("Invalid Signal")
```

#### 6.	user defined functions with for loop

```python
def print_n_number(n):
  for i in range(1,n+1):
    print(i)
```

```python
print_n_number(6)
```

```python
def sum_of_string_number(string):
  sum = 0
  for i in string:
    sum += int(i)
  return sum
```

```python
sum_of_string_number("12345")
```

#### 7.	user defined functions with for loop and if condition

```python
def print_even_or_odd(string):
  for i in string:
    if int(i) % 2 == 0:
      print(f"{i} - Even")
    else:
      print(f"{i} - Odd")
```

```python
print_even_or_odd("12345")
```

```python
def count_vowels(text):
  vowels = 0
  for i in text:
    if i in "aeiou":
      vowels += 1
  return vowels
      
```

```python
count_vowels("derin")
```

### 8.	break

```python
for i in range(1, 11):
  if i == 5:
    break
  print(i)
```

```python
while True:
    print(f"_MENU_\n1. Open\n2.Save\n3.Exit")
    user_input = int(input("Choose a option : "))
    
    if user_input == 1:
      print("--> File Opened")
    elif user_input == 2:
      print("--> File Saved")
    elif user_input == 3:
      print("--> Thank You, Bye")
      break
    else:
      print("--> Invalid Operation")
```

#### 9.	continue

```python
for i in range(1, 11):
  if i == 5 or i == 8:
    continue
  print(i)
```

```python
print("Even Numbers")
for i in range(1, 20):
  if i % 2 != 0:
    continue
  print(i)
```

#### 13.	if with in operator

```python
sentence = "This is content"

letter = input("Enter a letter: ")

if letter in sentence:
  print("Letter found")
```

```python
colors = "blue green yellow orange"
color = input("Enter a color: ")

if color in colors:
  print("Color Found")
```

#### 14.	if with not in operator

```python
sentence = "This is content"

letter = input("Enter a letter: ")

if letter not in sentence:
    print("Letter not found")
```

```python
colors = "blue green yellow orange"
color = input("Enter a color: ")

if color not in colors:
    print("Color not Found")
```
