# Task 1
- Give 2 examples of all String functions which should be meaningful.


---

#### 1. capitalize()

```python
name = "derin"
location = "bangalore"

name = name.capitalize()
location = location.capitalize()

print(name)
print(location)
```

#### 2. casefold()

```python
name = "DERIN"
location = "BaNGalore"

name = name.casefold()
location = location.casefold()

print(name)
print(location)
```

#### 3. lower()

```python
name = "DERIN"
location = "BaNGalore"

name = name.lower()
location = location.lower()

print(name)
print(location)
```

#### 4. upper()

```python
name = "derin"
location = "BaNGalore"

name = name.upper()
location = location.upper()

print(name)
print(location)
```

#### 5. title()

```python
name = "derin girish"
address = "banGLore, karNataKa"

name = name.title()
address = address.title()

print(name)
print(address)
```

#### 6. swapcase()

```python
name = "DeRiN"
location = "bAnGaLoRe"

name = name.swapcase()
location = location.swapcase()

print(name)
print(location)
```

#### 7. isalnum()

```python
code_1 = "derin123"
code_2 = "#%#^"

print(code_1.isalnum())
print(code_2.isalnum())
```

#### 8. isalpha()

```python
code_1 = "derin"
code_2 = "123"

print(code_1.isalpha())
print(code_2.isalpha())
```

#### 9. isascii()

```python
code_1 = "derin123@#$()"
code_2 = "derin😀😊"

print(code_1.isascii())
print(code_2.isascii())
```

#### 10. isdecimal()

```python
code_1 = "12335"
code_2 = "234.23"

print(code_1.isdecimal())
print(code_2.isdecimal())
```

#### 11. isdigit()

```python
code_1 = "123423"
code_2 = "derin"

print(code_1.isdigit())
print(code_2.isdigit())
```

#### 12. isnumeric()

```python
code_1 = "12323"
code_2 = "-123"

print(code_1.isnumeric())
print(code_2.isnumeric())
```

#### 13. islower()

```python
name = "derin"
location = "Bangalore"

print(name.islower())
print(location.islower())
```

#### 14. isupper()

```python
name = "DERIN"
location = "Bangalore"

print(name.isupper())
print(location.isupper())
```

#### 15. istitle()

```python
name = "Derin Girish"
location = "bangalore, Karnataka"

print(name.istitle())
print(location.istitle())
```

#### 16. isspace()

```python
name = "   "
location = "   Bangalore   "

print(name.isspace())
print(location.isspace())
```

#### 17. isprintable()

```python
name = "derin"
location = "\n"

print(name.isprintable())
print(location.isprintable())
```

#### 18. isidentifier()

```python
print("value_123".isidentifier())
print("123_value".isidentifier())
```

#### 19. find()

```python
text = "This is Derin Girish, Learning Data Science Course in PROITBRIDGE under guidence of AI Coach John"

print(text.find("Data"))
print(text.find("not"))
```

#### 20. index()

```python
text = "This is Derin Girish, Learning Data Science Course in PROITBRIDGE under guidence of AI Coach John"

print(text.index("u"))
print(text.index("not"))
```

#### 21. rindex()

```python
text = "This is Derin Girish, Learning Data Science Course in PROITBRIDGE under guidence of AI Coach John"

print(text.rindex("u"))
print(text.rindex("not"))
```

#### 22. count()

```python
text = "This is Derin Girish, Learning Data Science Course in PROITBRIDGE under guidence of AI Coach John"

print(text.count("u"))
print(text.count("not"))
```

#### 23. startswith()

```python
text = "Derin Girish"

print(text.startswith("Der"))
print(text.startswith("ish"))
```

#### 24. endswith()

```python
text = "Derin Girish"

print(text.endswith("ish"))
print(text.endswith("asd"))
```

#### 25. replace()

```python
text = "Derin Girish"
text_1 = "This is orange."

text = text.replace("eri", "abc")
text_1 = text_1.replace("orange", "apple")

print(text)
print(text_1)
```

#### 26. center()

```python
name = "Derin"
location = "bangalore"

print(f"-{name.center(10)}-")
print(location.center(20,"*"))
```

#### 27. ljust()

```python
name = "Derin"
location = "bangalore"

print(f"-{name.ljust(10)}-")
print(location.ljust(20, "*"))
```

#### 28. rjust()

```python
name = "Derin"
location = "bangalore"

print(f"-{name.rjust(10)}-")
print(location.rjust(20, "*"))
```

#### 29. zfill()

```python
price_1 = "20"
price_2 = "305"

print(price_1.zfill(10))
print(price_2.zfill(10))
```

#### 30. format()

```python
text = "My Name is {} and i am {} years old.".format("Derin Girish", 22)
text_1 = "My Name is {name} and i am {age} years old.".format(age = 22, name = "Derin Girish")

print(text)
print(text_1)
```

#### 31. format_map()

```python
data_dict = {"name" : "Derin", "age" : "12"}
data_text = "My Name is {name}, I am {age} years old.".format_map(data_dict)

client_dict = {"name" : "Girish", "location" : "Bangalore", "pricing" : 23}
client_text = "My Client name is {name}, and he is from {location} and the pricing is {pricing} lks".format_map(client_dict)

print(data_text)
print(client_text)
```

#### 32. strip()

```python
name = "   Derin   "
location = "  Bangalore"

print(f"-{name.strip()}-")
print(f"-{location.strip()}-")
```

#### 33. lstrip()

```python
name = "   Derin   "
location = "  Bangalore"

print(f"-{name.lstrip()}-")
print(f"-{location.lstrip()}-")
```

#### 34. rstrip()

```python
name = "   Derin   "
location = "Bangalore   "

print(f"-{name.rstrip()}-")
print(f"-{location.rstrip()}-")
```

#### 35. split()

```python
text = "Hi Everyone, This is Python Course"

split_text_1 = text.split(" ")
split_text_2 = text.split(" ", 2)

print(split_text_1)
print(split_text_2)
```

#### 36. rsplit()

```python
text = "Hi Everyone, This is Python Course"

split_text_1 = text.rsplit(" ")
split_text_2 = text.rsplit(" ", 2)

print(split_text_1)
print(split_text_2)
```

#### 37. splitlines()

```python
text = "Hi Everyone\n This is Python Course"

splitlines_text_1 = text.splitlines()
splitlines_text_2 = text.splitlines(True)

print(splitlines_text_1)
print(splitlines_text_2)
```

#### 38. join()

```python
my_list = ["Hi", "Everyone", ",", "Derin", "here"]

text = " ".join(my_list)
text_1 = "****".join(my_list)

print(text)
print(text_1)
```

#### 39. partition()

```python
text = "Derin here, How are you?"

partition_1 = text.partition("here")
partition_2 = text.partition("not")

print(partition_1)
print(partition_2)
```

#### 40. rpartition()

```python
text = "Derin here, How are you?"

partition_1 = text.rpartition("How")
partition_2 = text.rpartition("not")

print(partition_1)
print(partition_2)
```

#### 41. encode()

```python
text = "hai 👋"
text_1 = "å"

print(text.encode())
print(text_1.encode())
```

```python
text_1 = "Hai\t, How are you?"
text_2 = "I am \t Derin"

print(text_1.expandtabs(2))
print(text_2.expandtabs(8))
```
