## Word Guessing Game (Mini Hangman)
##### Objective:
Create a word-guessing game where the player tries to guess the letters of a hidden word. \
This will test your: \
●	Loops \
●	Conditional logic \
●	String manipulation \
●	Exception handling \
●	Logical thinking

Instructions:
1.	Create a list of random words (like “python”, “machine”, “learning”, “bridge”).
2.	Randomly select one word for the game.
3.	Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.
4.	Keep track of wrong guesses (max 6 attempts).
5.	End the game when the player either guesses the full word or uses all chances.


```python
import random

words = [
    "apple",
    "banana",
    "mango",
    "orange",
    "grapes",
    "pineapple",
    "watermelon",
    "papaya",
    "guava",
    "kiwi",
    "strawberry",
    "blueberry",
    "pear",
    "peach",
    "cherry",
    "pomegranate",
    "coconut",
    "lemon",
    "lychee",
    "dragonfruit",
]

print("Welcome...")

while True:
    word = random.choice(words)
    word = [i for i in word]
    print(f"Word: {' '.join(word)}")

    print("Your need to find : ")
    predicted_word = ["_" for _ in word]
    print(f"Predicatable Word Length: {' '.join(predicted_word)}")
    while True:
        try:
            chance = int(input("Enter number of chance Required : "))
            if chance <= 0:
                print("Chance Cannot be less than 1")
                continue
            break
        except ValueError:
            print("Enter a valid number")
            continue
        
    for i in range(1, chance + 1):
        while True:
            try:
                letter = input("Enter the letter: ")
                if len(letter) != 1:
                    raise ValueError("Type one letter.") 
                break
            except ValueError as e:
                print(e)
                continue
        for j in range(len(word)):
            if word[j] == letter:
                predicted_word[j] = letter
        print(f"Chance {i} - {' '.join(predicted_word)}")
        if predicted_word == word:
            print(f"You Found the Word in {i} Chance.")
            break
    if word == predicted_word:
        print(f"You Found the Word in {i} chance")
    else:
        print(f"You didn't found the word in {i} Chance")
    
    while True:
        try:
            option = input(f"Enter 'yes' to play, 'no' to stop....").lower()
            if option != "yes" and option != "no":
                raise ValueError("Please choose 'yes' or 'no'.")
            break
        except Exception as e:
            print(e)
            continue
    if option == "yes":
        continue
    elif option == "no":
        print("Thank You...")
        break
```
