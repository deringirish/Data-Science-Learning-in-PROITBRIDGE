## Rock, Paper, Scissors Game
#### Objective:
Build a simple Rock-Paper-Scissors game where the player competes with the computer. \
This task will test your understanding of: \
●	Input/output \
●	Conditional logic \
●	Random module \
●	Loops \
●	Exception handling 

#### Instructions:
1.	The user should enter their choice: rock, paper, or scissors.
2.	The computer should randomly select one of these.
3.	Compare both and print who wins.
4.	Ask if the player wants to play again- if yes, repeat; if no, exit.
5.	Handle invalid inputs gracefully using try-except or condition checks.


```python
import random

print(f"Welcome to Rock, Paper, Scissor Game......")

while True:
    options = ["Rock", "Paper", "Scissor"]
    computer_score = 0
    user_score = 0
    while True:
        try: 
            target_score = int(input("Enter the target score to win : "))
            if target_score <= 0:
                print("Target Value cannot be less than 0")
                continue
            break
        except ValueError:
            print("Invalid Input, Enter proper target value.")

    while True:
        computer_choice = random.choice(options)

        user_choice = input("Enter your choice: ").title()

        if user_choice not in options:
            print("Wrong letter/word entered! Please enter rock, paper, or scissor.")
            continue

        print(f"User - {user_choice} | Computer - {computer_choice}")

        if user_choice == computer_choice:
            print("Same Option - No point")
        elif (
            (user_choice == "Rock" and computer_choice == "Scissor")
            or (user_choice == "Paper" and computer_choice == "Rock")
            or (user_choice == "Scissor" and computer_choice == "Paper")
        ):
            user_score += 1
            print("User gets 1 point")
        else:
            computer_score += 1
            print("Computer gets 1 point")

        print("-"*20)

        print(f"User Score - {user_score}")
        print(f"Computer Score - {computer_score}")

        print("-"*20)

        if user_score >= target_score:
            print("User Won the Game")
            print("*" * 20)
            break

        if computer_score >= target_score:
            print("Computer won the game")
            print("*" * 20)
            break

    user_menu = input("Want to play again 'yes' or 'no': ").lower()
    if user_menu == "no":
        break
    else:
        continue

print("Thank you! See you again!")
```
