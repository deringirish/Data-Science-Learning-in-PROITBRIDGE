##### Parameters and functions

```python
addition(11, 11, 11) # Arguments - only duiring calling a particular function
```

#### Positional Arguments

```python
def get_details(name, place, expected_price):
  print(f"Client Name : {name}\nPlace : {place}\nExpected Price : {expected_price} Lakhs")
```

```python
get_details("Derin", "Bangalore", 9.7)
```

#### Keyword Arguments

```python
get_details(place="Bangalore", expected_price = 45.3, name = "Derin")
```

#### Default Parameters and Arguments

```python
def get_details_chennai(name, expected_price, place="Chennai"):
    print(
        f"Client Name : {name}\nPlace : {place}\nExpected Price : {expected_price} Lakhs"
    )
```

```python
get_details_chennai("Derin", 34.8)
```

#### Variable-lenth Argument

```python
def get_details_chennai(name,*expected_price, place="Chennai"):
    print(
        f"Client Name : {name}\nPlace : {place}\nExpected Price : {expected_price} Lakhs"
    )
```

```python
get_details_chennai("Derin", 34.8, 45, 35)
```

```python
def get_details_chennai(name,*sample, place="Chennai", **expected_price):
    print(
        f"Client Name : {name}\nPlace : {place}\nExpected Price : {expected_price} Lakhs\nSample : {sample}"
    )
```

```python
get_details_chennai(
    "Derin",
    23,
    56,
    45,
    ask_1=34.8,
    ask_2=45,
    ask_3=35,
)
```

### 5.3 Control the Flow of the program

```python
user_password = input("New Password")
print(len(user_password))
if len(user_password) < 5:
    print("WEAK Password")
elif 5 <= len(user_password) <= 9:
  print("MODERATE Password")
else:
    print("STRONG Password")
```

### 5.4 Challenge -2 Find the Factors of a Numbers

```python
num = int(input("Enter the number: "))
result = []
for i in range(1, num+1):
    if num % i == 0:
        result.append(f"{num} / {i} = {num/i}")
        # print(f"{num} / {i} = {num/i}")20
for i in result:
  print(i)
```
