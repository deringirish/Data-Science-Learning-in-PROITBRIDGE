```python
import pandas as pd

df = pd.read_csv("../data/full_sales_data.csv")

# Step 1: keep needed columns
df = df[["Date", "Voucher", "Party", "Product", "Qty", "Rate"]]

# Step 2: remove missing rows
df = df.dropna()

# Step 3: remove spaces
for col in df.columns:
    df[col] = df[col].astype(str).str.strip()

# Step 4: remove empty rows
df = df[(df != "").all(axis=1)]

# Step 5: clean Voucher
df["Voucher"] = df["Voucher"].str.extract("(\d+)")

# Step 6: clean numeric columns
df["Qty"] = pd.to_numeric(df["Qty"].str.replace(",", ""), errors="coerce")
df["Rate"] = pd.to_numeric(df["Rate"].str.replace(",", ""), errors="coerce")

# Step 7: remove invalid rows
df = df.dropna()

# Step 8: save cleaned file
# df.to_csv("cleaned_sales.csv", index=False)
```
