```python
profil_and_loss = cola_data.iloc[2:20, 1:]

profil_and_loss = profil_and_loss.T
profil_and_loss.columns = [
    "Financial Year",
    "NET OPERATING REVENUES",
    "Cost of goods sold",
    "Gross Profit",
    "Selling, general and administrative expenses",
    "Other operating charges",
    "Operating Income",
    "Interest income",
    "Interest expense",
    "Equity income (loss) - net",
    "Other income (loss) - net",
    "Income before income taxes",
    "Income taxes",
    "CONSOLIDATED NET INCOME",
    "Less: Net income attributable to noncontrolling interests",
    "Net income from continuing operations",
    "Income (Loss) from Discontinued Operations, Net of Tax, Including Portion Attributable to Noncontrolling Interest",
    "NET INCOME ATTRIBUTABLE TO SHAREOWNERS OF THE COCA-COLA COMPANY",
]
profil_and_loss.reset_index(drop=True, inplace=True)
profil_and_loss = profil_and_loss.iloc[1:].reset_index(drop=True)
profil_and_loss
```

```python
profil_and_loss.columns
```

```python
cols = ['NET OPERATING REVENUES', 'Cost of goods sold',
       'Gross Profit', 'Selling, general and administrative expenses',
       'Other operating charges', 'Operating Income', 'Interest income',
       'Interest expense', 'Equity income (loss) - net',
       'Other income (loss) - net', 'Income before income taxes',
       'Income taxes', 'CONSOLIDATED NET INCOME',
       'Less: Net income attributable to noncontrolling interests',
       'Net income from continuing operations',
       'Income (Loss) from Discontinued Operations, Net of Tax, Including Portion Attributable to Noncontrolling Interest',
       'NET INCOME ATTRIBUTABLE TO SHAREOWNERS OF THE COCA-COLA COMPANY']
profil_and_loss[cols] = profil_and_loss[cols].apply(pd.to_numeric).astype("Int64")
```

```python
profil_and_loss.rename(
    columns={
        "NET OPERATING REVENUES": "Revenue",
        "Cost of Goods Sold": "COGS",
        "Gross Profit": "Gross_Profit",
        "Selling, General and Administrative Expenses": "SGA",
        "Other Operating Charges": "Other_Op_Charges",
        "Operating Income": "Operating_Income",
        "Interest Income": "Interest_Income",
        "Interest Expense": "Interest_Expense",
        "Equity Income (Loss) — Net": "Equity_Income",
        "Other Income (Loss) — Net": "Other_Income",
        "Income Before Income Taxes": "PreTax_Income",
        "Income Taxes": "Taxes",
        "CONSOLIDATED NET INCOME": "Net_Income",
        "Less: Net income attributable to noncontrolling interests": "NonControl_Int",
        "Net income from continuing operations": "Cont_Income",
        "Income (Loss) from Discontinued Operations, Net of Tax, Including Portion Attributable to Noncontrolling Interest": "Disc_Income",
        "NET INCOME ATTRIBUTABLE TO SHAREOWNERS OF THE COCA-COLA COMPANY": "Shareholder_Income",
    },
    inplace=True,
)
```

### Gross Profit
profil_and_loss["Gross_Profit"] = (
    profil_and_loss["Revenue"]
    - profil_and_loss["COGS"]
)

### Operating Income
profil_and_loss["Operating_Income"] = (
    profil_and_loss["Gross_Profit"]
    - profil_and_loss["SGA"]
    - profil_and_loss["Other_Op_Charges"]
)

### Net Interest
profil_and_loss["Net_Interest"] = (
    profil_and_loss["Interest_Income"]
    - profil_and_loss["Interest_Expense"]
)

### Non Operating Income
profil_and_loss["Non_Operating_Income"] = (
    profil_and_loss["Interest_Income"]
    - profil_and_loss["Interest_Expense"]
    + profil_and_loss["Equity_Income"]
    + profil_and_loss["Other_Income"]
)

### PreTax Income
profil_and_loss["PreTax_Income"] = (
    profil_and_loss["Operating_Income"]
    + profil_and_loss["Non_Operating_Income"]
)

### Net Income
profil_and_loss["Net_Income"] = (
    profil_and_loss["PreTax_Income"]
    - profil_and_loss["Taxes"]
)

### Continuing Income
profil_and_loss["Cont_Income"] = (
    profil_and_loss["Net_Income"]
    - profil_and_loss["Disc_Income"].fillna(0)
)

### Shareholder Income
profil_and_loss["Shareholder_Income"] = (
    profil_and_loss["Net_Income"]
    - profil_and_loss["NonControl_Int"]
)

```python
import pandas as pd
import numpy as np

# Formula dictionary using YOUR columns
formulas = {
    "Gross_Profit": lambda r: r["Revenue"] - r["Cost of goods sold"],
    "Operating_Income": lambda r: r["Gross_Profit"]
    - r["Selling, general and administrative expenses"]
    - r["Other operating charges"],
    "Income before income taxes": lambda r: r["Operating_Income"]
    + r["Interest income"]
    - r["Interest expense"]
    + r["Equity income (loss) - net"]
    + r["Other income (loss) - net"],
    "Net_Income": lambda r: r["Income before income taxes"] - r["Income taxes"],
    "Cont_Income": lambda r: r["Net_Income"]
    - (0 if pd.isna(r["Disc_Income"]) else r["Disc_Income"]),
    "Shareholder_Income": lambda r: r["Net_Income"] - r["NonControl_Int"],
}

failed_records = []

# Row-by-row validation
for idx, row in profil_and_loss.iterrows():

    for formula_name, formula in formulas.items():

        try:

            expected = formula(row)
            actual = row[formula_name]

            # Skip if both missing
            if pd.isna(actual) and pd.isna(expected):
                continue

            # Safe comparison
            if not np.isclose(float(actual), float(expected), rtol=1e-5):

                failed_records.append(
                    {
                        "Index": idx,
                        "FY": row["Financial Year"],
                        "Formula": formula_name,
                        "Actual": actual,
                        "Expected": expected,
                    }
                )

        except Exception as e:

            failed_records.append(
                {
                    "Index": idx,
                    "FY": row["Financial Year"],
                    "Formula": formula_name,
                    "Error": str(e),
                }
            )

# Print only failed validations
if len(failed_records) == 0:
    print("All formulas validated successfully.")

else:
    for fail in failed_records:

        print(
            f"Row={fail['Index']} | "
            f"FY={fail['FY']} | "
            f"Formula={fail['Formula']} FAILED | "
            f"Actual={fail.get('Actual')} | "
            f"Expected={fail.get('Expected')}"
        )
```
