````python
profil_and_loss = cola_data.iloc[2:20, 1:]

profil_and_loss = profil_and_loss.T
profil_and_loss.columns = [
    "Financial Year",
    "NET OPERATING REVENUES",
    "Cost of goods sold",
    "Gross Profit",
    "Selling, general and administrative expenses",
    "Other operating charges",
    "Operating Income",
    "Interest income",
    "Interest expense",
    "Equity income (loss) - net",
    "Other income (loss) - net",
    "Income before income taxes",
    "Income taxes",
    "CONSOLIDATED NET INCOME",
    "Less: Net income attributable to noncontrolling interests",
    "Net income from continuing operations",
    "Income (Loss) from Discontinued Operations, Net of Tax, Including Portion Attributable to Noncontrolling Interest",
    "NET INCOME ATTRIBUTABLE TO SHAREOWNERS OF THE COCA-COLA COMPANY",
]
profil_and_loss.reset_index(drop=True, inplace=True)
profil_and_loss = profil_and_loss.iloc[1:].reset_index(drop=True)
profil_and_loss
````

````python
cols = ['NET OPERATING REVENUES', 'Cost of goods sold',
       'Gross Profit', 'Selling, general and administrative expenses',
       'Other operating charges', 'Operating Income', 'Interest income',
       'Interest expense', 'Equity income (loss) - net',
       'Other income (loss) - net', 'Income before income taxes',
       'Income taxes', 'CONSOLIDATED NET INCOME',
       'Less: Net income attributable to noncontrolling interests',
       'Net income from continuing operations',
       'Income (Loss) from Discontinued Operations, Net of Tax, Including Portion Attributable to Noncontrolling Interest',
       'NET INCOME ATTRIBUTABLE TO SHAREOWNERS OF THE COCA-COLA COMPANY']
profil_and_loss[cols] = profil_and_loss[cols].apply(pd.to_numeric).astype("Int64")
````

### Formula - Calculation Heppen in this Data

| Feature                                                                                                               | Type                | Formula                                                                                            |
| --------------------------------------------------------------------------------------------------------------------- | ------------------- | -------------------------------------------------------------------------------------------------- |
| Financial Year                                                                                                        | Provided            | —                                                                                                  |
| NET OPERATING REVENUES                                                                                                | Provided            | —                                                                                                  |
| Cost of goods sold                                                                                                    | Provided            | —                                                                                                  |
| **Gross Profit**                                                                                                      | Calculated          | NET OPERATING REVENUES − Cost of goods sold                                                        |
| Selling, general and administrative expenses                                                                          | Provided            | —                                                                                                  |
| Other operating charges                                                                                               | Provided            | —                                                                                                  |
| **Operating Income**                                                                                                  | Calculated          | Gross Profit − Selling, general and administrative expenses − Other operating charges              |
| Interest income                                                                                                       | Provided            | —                                                                                                  |
| Interest expense                                                                                                      | Provided            | —                                                                                                  |
| Equity income (loss) - net                                                                                            | Provided            | —                                                                                                  |
| Other income (loss) - net                                                                                             | Provided            | —                                                                                                  |
| **Income before income taxes**                                                                                        | Calculated          | Operating Income + Interest income − Interest expense + Equity income (loss) + Other income (loss) |
| Income taxes                                                                                                          | Provided            | —                                                                                                  |
| **Net income from continuing operations**                                                                             | Calculated          | Income before income taxes − Income taxes                                                          |
| **Income (Loss) from Discontinued Operations, Net of Tax, Including Portion Attributable to Noncontrolling Interest** | Provided / Reported | —                                                                                                  |
| **CONSOLIDATED NET INCOME**                                                                                           | Calculated          | Net income from continuing operations + Income (Loss) from Discontinued Operations                 |
| Less: Net income attributable to noncontrolling interests                                                             | Provided            | —                                                                                                  |
| **NET INCOME ATTRIBUTABLE TO SHAREOWNERS OF THE COCA-COLA COMPANY**                                                   | Calculated          | CONSOLIDATED NET INCOME − Less: Net income attributable to noncontrolling interests                |


````python
profil_and_loss["Net income from continuing operations"] = (
    profil_and_loss["Income before income taxes"] - profil_and_loss["Income taxes"]
)

profil_and_loss[
    "Income (Loss) from Discontinued Operations, Net of Tax, Including Portion Attributable to Noncontrolling Interest"
] = profil_and_loss[
    "Income (Loss) from Discontinued Operations, Net of Tax, Including Portion Attributable to Noncontrolling Interest"
].fillna(0)
````

````python
import pandas as pd
import numpy as np

# Create separate validation dataframe
validation_df = pd.DataFrame()

# Optional: keep identifier column
validation_df['Financial Year'] = profil_and_loss['Financial Year']

# Validation checks
validation_df['Gross_Profit_valid'] = np.isclose(
    profil_and_loss['Gross Profit'],
    profil_and_loss['NET OPERATING REVENUES'] - profil_and_loss['Cost of goods sold'],
    equal_nan=True
)

validation_df['Operating_Income_valid'] = np.isclose(
    profil_and_loss['Operating Income'],
    profil_and_loss['Gross Profit']
    - profil_and_loss['Selling, general and administrative expenses']
    - profil_and_loss['Other operating charges'],
    equal_nan=True
)

validation_df['Income_before_taxes_valid'] = np.isclose(
    profil_and_loss['Income before income taxes'],
    profil_and_loss['Operating Income']
    + profil_and_loss['Interest income']
    - profil_and_loss['Interest expense']
    + profil_and_loss['Equity income (loss) - net']
    + profil_and_loss['Other income (loss) - net'],
    equal_nan=True
)

validation_df['Continuing_Operations_valid'] = np.isclose(
    profil_and_loss['Net income from continuing operations'],
    profil_and_loss['Income before income taxes']
    - profil_and_loss['Income taxes'],
    equal_nan=True
)

validation_df['Consolidated_Net_Income_valid'] = np.isclose(
    profil_and_loss['CONSOLIDATED NET INCOME'],
    profil_and_loss['Net income from continuing operations']
    + profil_and_loss['Income (Loss) from Discontinued Operations, Net of Tax, Including Portion Attributable to Noncontrolling Interest'],
    equal_nan=True
)

validation_df['Shareowners_Income_valid'] = np.isclose(
    profil_and_loss['NET INCOME ATTRIBUTABLE TO SHAREOWNERS OF THE COCA-COLA COMPANY'],
    profil_and_loss['CONSOLIDATED NET INCOME']
    - profil_and_loss['Less: Net income attributable to noncontrolling interests'],
    equal_nan=True
)

# Overall validation
validation_cols = validation_df.columns.drop('Financial Year')

validation_df['All_Valid'] = validation_df[validation_cols].all(axis=1)
````
