```python
import pandas as pd
```

```python
concert_data = pd.read_csv("data/my_file.csv")
concert_data
```

```python
concert_data.columns
```

```python
concert_data.columns.tolist()
```

```python
concert_data = concert_data.rename(
    columns={
        "Actual\xa0gross": "Actual gross",
        "Adjusted\xa0gross (in 2022 dollars)": "Adjusted gross (in 2022 dollars)",
    }
)
```

```python
concert_data = concert_data[
    [
        "Rank",
        "Peak",
        "All Time Peak",
        "Actual gross",
        "Adjusted gross (in 2022 dollars)",
        "Artist",
        "Tour title",
        "Year(s)",
        "Shows",
        "Average gross",
    ]
]
```

```python
concert_data.info()
```

```python
concert_data.head()
```

```python
concert_data["Peak"] = pd.to_numeric(
    concert_data["Peak"]
    .astype(str)
    .replace("\[([^\]]+)\]", "", regex=True)
    .astype("Int64")
)

concert_data["All Time Peak"] = pd.to_numeric(
    concert_data["All Time Peak"]
    .astype(str)
    .replace("\[([^\]]+)\]", "", regex=True)
    .astype("Int64")
)

concert_data["Actual gross"] = pd.to_numeric(
    concert_data["Actual gross"]
    .astype(str)
    .str.replace("$", "")
    .str.replace(",", "")
    .str.replace("\[([^\]]+)\]", "", regex=True)
    .astype(float)
)

concert_data["Adjusted gross (in 2022 dollars)"] = pd.to_numeric(
    concert_data["Adjusted gross (in 2022 dollars)"]
    .astype(str)
    .str.replace("$", "")
    .str.replace(",", "")
    .str.replace("\[([^\]]+)\]", "", regex=True)
    .astype(float)
)

concert_data["Artist"] = concert_data["Artist"].str.strip().str.upper()


concert_data["Tour title"] = (
    concert_data["Tour title"]
    .str.strip()
    .replace("^\[[^\]]+\]|\[[^\]]+\]$", "", regex=True)
    .str.replace("[^a-zA-Z0-9\s]", "", regex=True)
    .str.title()
)


concert_data["Start year"] = (
    concert_data["Year(s)"].astype(str).str.extract(r"(^\d+)")[0].astype("Int64")
)

concert_data["End year"] = (
    concert_data["Year(s)"].astype(str).str.extract(r"(\d+$)")[0].astype("Int64")
)

concert_data["Tour duration"] = concert_data["End year"] - concert_data["Start year"] + 1
concert_data = concert_data.drop("Year(s)", axis=1)

concert_data["Average gross"] = pd.to_numeric(
    concert_data["Average gross"]
    .astype(str)
    .str.replace("$", "")
    .str.replace(",", "")
    .str.replace("\[([^\]]+)\]", "", regex=True)
    .astype(float)
)
```
