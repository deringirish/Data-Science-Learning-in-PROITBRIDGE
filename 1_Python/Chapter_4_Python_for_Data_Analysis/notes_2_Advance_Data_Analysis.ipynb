#### Discrete and Continous Data
- Discrete -> Just Countable / Can't be segmented
- Continous -> Mesurable / Have units / Can also be segemented

---

**Discrete data** and **Continuous data** are two types of **numerical data** in statistics and data analysis.

Here's the difference:

| Feature         | Discrete Data                   | Continuous Data             |
| --------------- | ------------------------------- | --------------------------- |
| **Definition**  | Countable values                | Measurable values           |
| **Values**      | Specific, separate values       | Any value within a range    |
| **Decimals?**   | Usually no                      | Yes                         |
| **Obtained by** | Counting                        | Measuring                   |
| **Examples**    | Number of students, cars, books | Height, weight, temperature |

#### 1. Discrete Data

Discrete data consists of **distinct/countable values**.

You **count** discrete data.

##### Examples:

* Number of students in a class → 40, 41, 42
* Number of employees → 10, 15
* Number of calls received
* Number of children in a family

You cannot have:

* 3.5 students ❌
* 2.7 cars ❌

So these are **whole-number counts**.

##### Example in Python:

```python id="u4t1x5"
students = [30, 25, 40, 35]
```


#### 2. Continuous Data

Continuous data can take **any value within a range**.

You **measure** continuous data.

##### Examples:

* Height → 170.2 cm
* Weight → 65.8 kg
* Salary → 55000.75
* Temperature → 36.7°C
* Time taken → 2.35 sec

These can have decimal values.

##### Example in Python:

```python id="g15y7y"
height = [165.5, 170.2, 168.9]
```

### 3. Aggregate

````python
insurance_data.head()
````

#### 3.1 Find out the average insurance charges based on region

````python
## This is only by using groupby function
insurance_data.groupby(by="region")["charges"].mean().sort_values(), round(2)
````

````python
insurance_data.groupby(by="region")["charges"].agg(func=["mean", "min", "max", "sum"]).round(2)
````

#### 3.2 Find out the average insurance charges based on region

````python
insurance_data.groupby(by=["region", "sex"])[["charges"]].agg(func=["mean", "min", "max", "sum"]).round(2)
````

### 4. Cross Tab

crosstab() in pandas is used to count the frequency (occurrence) between two or more categorical/discrete variables.

````python
insurance_data.head()
````

#### 4.1 How many smokers and non-smoker based on region

````python
pd.crosstab(index=insurance_data["region"], columns=insurance_data["smoker"])
````

````python
pd.crosstab(
    index=insurance_data["region"], columns=insurance_data["smoker"], margins=True
)
````

````python
pd.crosstab(index=[insurance_data["region"], insurance_data["sex"]], columns=insurance_data["smoker"], margins=True)
````

````python
pd.crosstab(index=insurance_data['region'] , columns=[insurance_data["smoker"], insurance_data['sex']], margins=True)
````
