## Conditional Filtering

### 4.2 List the Teams which has got Red Cards

#### Pandas Filtering Notes

``` python
euro12[euro12["Goals"] > 5][["Team", "Goals"]]
```

1.  `euro12["Goals"] > 5` → Creates Boolean mask (True/False).
2.  `euro12[mask]` → Filters rows where condition is True.
3.  `[["Team","Goals"]]` → Selects required columns.
4.  Flow: Mask → Filter → Select.
5.  Cleaner version:

``` python
mask = euro12["Goals"] > 5
result = euro12[mask][["Team","Goals"]]
```
