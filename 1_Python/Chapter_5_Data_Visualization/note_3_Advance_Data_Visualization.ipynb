# Advance Level of Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
import plotly.express as px

df = px.data.gapminder().query("year == 2007").query("continent == 'Europe'")
df.loc[df["pop"] < 2.0e6, "country"] = (
    "Other countries"  # Represent only large countries
)
fig = px.pie(
    df, values="pop", names="country", title="Population of European continent"
)
fig.show()
```

```python
mtcars_data = pd.read_csv("data/mtcars.csv")
```

```python
sns.histplot(mtcars_data, x="mpg", hue="am", multiple="dodge", kde=True)
```

```python
penguins = sns.load_dataset("penguins")
sns.histplot(
    data=penguins,
    x="flipper_length_mm",
    # hue="species",
    # multiple="stack",
    kde=True,
    bins=20,
    color="skyblue"
    # palette="viridis",
)
plt.title("Penguin Flipper Length Distribution by Species")

plt.show()
```
