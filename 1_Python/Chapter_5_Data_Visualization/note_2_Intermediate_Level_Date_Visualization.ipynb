# Intermediate Level Data Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
mtcars_data = pd.read_csv("data/mtcars.csv")
mtcars_data.head(10)
```

## Must Know 
- should have the clarify between **Discrete & Continious data**

### Discrete vs Continuous
| Feature          | Discrete         | Continuous        |
| ---------------- | ---------------- | ----------------- |
| Counted?         | ✅ Yes            | ❌ No (measured)   |
| Measured?        | ❌ Usually no     | ✅ Yes             |
| Decimal values?  | ❌ Usually no     | ✅ Yes             |
| Separate values? | ✅ Yes            | ❌ No              |
| Examples         | Cylinders, gears | Weight, time, MPG |

## 1. Univariate Analysis
---
**Important Charts**
1. Histogram
2. Countplot
4. Pie chart

#### 1.1 Histogram
- To understand the distribution of one **Continious Data**

```python
plt.hist(mtcars_data.mpg)
plt.xlabel("MPG")
plt.ylabel("Frequency")
plt.title("MPG Distribution")
plt.show()
```

#### 1.2 Countplot
- Counting the Number of Occurances
- To understand the distribution of one **Discrete Data**
- can pass the value of x or y, can't pass both at a time

```python
sns.countplot(data=mtcars_data, x="am")
plt.title("Automatic vs Manual Car Distribution")
plt.show()
```

#### Static Chat vs Dynamic Chat
- Static Chat - with static Value
- Dynamic Chat - Data change Dynamically

#### 1.3 Pie Chart
- Data should need to be **Discrete Data**

```python
mtcars_data.cyl.value_counts().to_dict()
```

```python
# plt.pie(data=mtcars_data, x="mpg")
data = mtcars_data.cyl.value_counts()
plt.figure(figsize=(5, 5))
plt.pie(
    x=data,
    labels=[f"{i} cyl" for i in data.index],
    explode=[0.1] * len(data),
    autopct="%1.1f%%",
)
plt.show()
```

#### Correlation Function
- To see the stength of relatioship between 2 featues

##### Types of correlation
1. Pearson Correlation Co-efficient ( important )
2. Spearman's Correlation Co-efficient

##### 1. Pearson Correlation Co-efficient
- the values range from -1 to 1
- when x increase and y also increase positively correlated
- when y increase and x also increase negatively correlated
    - -1 - 0 is negatively correlated
    - 0 - 1 is positively correlated
    - 0 is no correlated

- 0 to 0.20 = Very Weak Corrlation
- 0.21 to 0.40 = Weak Correlationo
- 0.41 to 0.60 = Normal/Standard Correlation
- 0.61 to 0.80 = Strong Correlation
- 0.81 to 1 = Very Strong Correlation

```python
del mtcars_data["model"]
```

```python
corr_mat = mtcars_data.corr(method='pearson').round(2)
corr_mat
```

```python
plt.figure(figsize=(10,10))
sns.heatmap(corr_mat, annot=True)
```

#### Insights
1. x - mpg, y - cyl
