# Handling Imbalance Dataset

- There are several types to to this 
    1. tweaking **Stratify** in **train_test_split**
        - Eg| `train_test_split(X, y, test_size=0.2, random_state=123, stratify=y)`
    2. tweaking **class_weight** in **ML Model algorithm**
        - Eg| `DecisionTreeClassifier(criterion="gini", max_depth=3, class_weight={0:2, 1:1})`
    3. Under Head Sampling
    4. Over Head Sampling
    5. Clustering Sampling

## 1. Import Necessay Libraries

```python
from sklearn import datasets
import pandas as pd
```

## 2. Import Dataset

```python
breast_cancer_data = datasets.load_breast_cancer()
```

```python
print(breast_cancer_data.DESCR)
```

```python
breast_cancer_data.feature_names
```

```python
breast_cancer_data_df = pd.DataFrame(data=breast_cancer_data.data, columns=breast_cancer_data.feature_names)
```

```python
breast_cancer_data_df["target"] = breast_cancer_data.target
```

```python
breast_cancer_data_df.target.value_counts()
```

## 5. Model Building

```python
X = breast_cancer_data_df.drop(labels=["target"], axis=1)
X.head()
```

```python
y = breast_cancer_data_df["target"]
y.head()
```

```python
print(f"X_Shape: {X.shape}")
print(f"y_Shape: {y.shape}")
```

```python
import matplotlib.pyplot as plt
```

```python
plt.pie(x = y.value_counts(), labels=["Benign", "Malignant"], autopct="%1.0f%%")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=123, stratify=y)
```

```python
print(f"Train Data Shape: {X_train.shape}, {y_train.shape}")
print(f"Test Data Shape: {X_test.shape}, {y_test.shape}")
```

```python
print(f"Train Data Value Count:\n {round(y_train.value_counts(normalize=True)*100, 3)}")
print("-----------")
print(f"Test Data Value Count:\n {round(y_test.value_counts(normalize=True)*100, 3)}")
```

```python
plt.pie(x = y_train.value_counts(), labels=["Benign", "Malignant"], autopct="%1.0f%%")
plt.show()
```

```python
plt.pie(x = y_test.value_counts(), labels=["Benign", "Malignant"], autopct="%1.0f%%")
plt.show()
```

## 6. Model Training

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt_model = DecisionTreeClassifier(criterion="gini", max_depth=3, class_weight={0:2, 1:1})
```

```python
dt_model.fit(X_train, y_train)
```

## 7. Model Testing and 

```python
y_pred_train = dt_model.predict(X_train)
```

```python
y_pred_test = dt_model.predict(X_test)
```

## 8. Model Evaluation

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

#### 8.1 Train Data

```python
print(classification_report(y_train, y_pred_train))
```

```python
print(confusion_matrix(y_train, y_pred_train))
```

#### 8.2 Test Data

```python
print(classification_report(y_test, y_pred_test))
```

```python
print(confusion_matrix(y_test, y_pred_test))
```
