### Feature Engineering

Whatever you do with features, we call it as Feature Engineering.

* Feature Elimination - dropping the features.
* Feature Addition - adding some features.
* Feature Transformation - transforming the given feature values into an another scale - Log Tranformation, Sqrt Transformation ..
* Feature Selection - deciding which features are important out of many features and choosing that features for model building.

#### Feature Selection Techniques

* sklearn - SelectFromModel
* sklearn - RFE(ie,Recursive Feature Elimination)

# 1. Importing Necessary Libraries

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import RFE, SelectFromModel

from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix

from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

import warnings
warnings.filterwarnings("ignore")
```

# 2. Import Dataset

```python
load_breast_cancer()
```

```python
data = load_breast_cancer()

bc_data = pd.DataFrame(data["data"], columns=data.feature_names)
bc_data["target"] = data["target"]
bc_data.sample(5)
```

# 3. Data Understanding

```python
bc_data.describe(include="all")
```

```python
X = bc_data.drop(labels=["target"], axis=1)
X
```

```python
y = bc_data.target
y
```

# 4. Model Building

```python
X_train, X_test, y_train, y_test  = train_test_split(X, y, test_size=0.2, shuffle=True, stratify=y, random_state=123)
```

```python
print(f"Train: {X_train.shape, y_train.shape}")
print(f"Test: {X_test.shape, y_test.shape}")
```

# 5. Model Training || 6. Model Testing || 7. Model Evaluation

```python
def run_random_forest_classifier(X_train, X_test, y_train, y_test):
    rf_model = RandomForestClassifier(max_depth=7, random_state=123)
    rf_model.fit(X_train, y_train)
    y_pred_test = rf_model.predict(X_test)

    print(f"Accuracy Score: {accuracy_score(y_test, y_pred_test)*100:.2f}%")
    print(f"Precision Score: {precision_score(y_test, y_pred_test)*100:.2f}%")
    print(f"Recall Score: {recall_score(y_test, y_pred_test)*100:.2f}%")
    print(f"Confusion Matrix: \n{confusion_matrix(y_test, y_pred_test)}")
```

```python
%%time
run_random_forest_classifier(X_train, X_test, y_train, y_test)
```

### How to choose the optimal number of featues to train my model

```python
# max_acc = {"i" : 0, "acc" : 0, "prec" : 0, "recall" : 0}
# max_prec = {"i" : 0, "acc" : 0, "prec" : 0, "recall" : 0}
# max_recall = {"i" : 0, "acc" : 0, "prec" : 0, "recall" : 0}
```

```python
def run_random_forest_classifier_with_max_option(X_train, X_test, y_train, y_test):
    rf_model = RandomForestClassifier(max_depth=7, random_state=123)
    rf_model.fit(X_train, y_train)
    y_pred_test = rf_model.predict(X_test)
    acc_score = accuracy_score(y_test, y_pred_test)
    prec_score = precision_score(y_test, y_pred_test)
    re_score = recall_score(y_test, y_pred_test)
    

    print(f"Accuracy Score: {acc_score*100:.2f}%")
    print(f"Precision Score: {prec_score*100:.2f}%")
    print(f"Recall Score: {re_score*100:.2f}%")
    print(f"Confusion Matrix: \n{confusion_matrix(y_test, y_pred_test)}")
    
    return acc_score, prec_score, re_score
    
```

```python
for i in range(len(X_train.columns)):
    rfe_model = RFE(estimator=GradientBoostingClassifier(), n_features_to_select=None)
    rfe_model.fit(X_train, y_train)
    X_train_rfe = rfe_model.transform(X_train)
    X_test_rfe = rfe_model.transform(X_test)
    print(f"Selected Featue: {i}:")
    acc_score, prec_score, re_score = run_random_forest_classifier_with_max_option(X_train_rfe, X_test_rfe, y_train, y_test)
    
    
    # if acc_score > max_acc["acc"]:
    #     max_acc["i"] = i+1
    #     max_acc["acc"] = acc_score
    #     max_acc["prec"] = prec_score
    #     max_acc["recall"] = re_score
    # if prec_score > max_prec["prec"]:
    #     max_prec["i"] = i+1
    #     max_prec["acc"] = acc_score
    #     max_prec["prec"] = prec_score
    #     max_prec["recall"] = re_score       
    # if re_score > max_recall["recall"]:
    #     max_recall["i"] = i+1
    #     max_recall["acc"] = acc_score
    #     max_recall["prec"] = prec_score
    #     max_recall["recall"] = re_score
    print(f"**********************************************************\n")
    
```
