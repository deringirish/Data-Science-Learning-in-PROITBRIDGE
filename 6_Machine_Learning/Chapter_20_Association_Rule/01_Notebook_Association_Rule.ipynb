##### Lets build best `ASSOCIATES` for `UK` as it has more transactions

```python
uk_data = online_retail_data[online_retail_data.Country == "United Kingdom"]
uk_data
```

```python
uk_data.isna().sum()
```

```python
print(f"Precentage of NaN data w.r.t Description: {(1454 / 495478) * 100:.2}%")
```

so it is less than 1 percentage so we can remove it

```python
del uk_data["CustomerID"]
```

```python
uk_data = uk_data.dropna()
```

```python
uk_data.isna().sum()
```

```python
uk_pivoted_data = pd.pivot_table(
    data=uk_data,
    values="Quantity",
    index="InvoiceNo",
    columns="Description",
    aggfunc="sum",
    fill_value=0,
).gt(0).astype(int)
```

```python
uk_pivoted_data.sample(20)
```

# 4. Data Mining

```python
freqent_item_sets = apriori(df=uk_pivoted_data, min_support=0.02, use_colnames=True)
```

```python
best_associates = association_rules(df=freqent_item_sets, metric="confidence", min_threshold=0.02)
```

```python
best_associates.to_csv("data/best_anticidents_consequence.csv")
```
