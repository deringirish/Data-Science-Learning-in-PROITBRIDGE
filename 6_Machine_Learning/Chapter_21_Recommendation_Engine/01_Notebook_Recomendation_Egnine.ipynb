## Building UBCF using Euclideon Distance

```python
from sklearn.metrics import pairwise_distances
```

```python
euclidean_pivot_table = pd.pivot_table(data=movie_data, values="rating", index="userId", columns="movie", fill_value=0)
euclidean_pivot_table
```

```python
euclidean_correlation = pd.DataFrame(
    pairwise_distances(X=euclidean_pivot_table, metric="euclidean"),
    index=euclidean_pivot_table.index,
    columns=euclidean_pivot_table.index,
)
euclidean_correlation.head(20)
```

```python
df = euclidean_correlation

ratings = df.stack()

# Exclude zero values
ratings = ratings[ratings > 0]

# Minimum 10
min_values = ratings.nsmallest(10)

min_values = pd.DataFrame({
    "userId": min_values.index.get_level_values(0),
    "movie": min_values.index.get_level_values(1),
    "rating": min_values.values
})

# Maximum 10
max_values = ratings.nlargest(10)

max_values = pd.DataFrame({
    "userId": max_values.index.get_level_values(0),
    "movie": max_values.index.get_level_values(1),
    "rating": max_values.values
})
```

```python
display(min_values)
display(max_values)
```

## Bulding UBCF using consine similarity

```python
cosine_correlation = pd.DataFrame(
    pairwise_distances(X=euclidean_pivot_table, metric="cosine"),
    index=euclidean_pivot_table.index,
    columns=euclidean_pivot_table.index,
)
cosine_correlation.head(20)
```
